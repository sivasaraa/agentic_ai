```python
import chromadb

client = chromadb.Client()

collection = client.create_collection("company_articles")
```

```python
collection.add(
    ids=["id1","id2","id3","id4"],
    documents=[
        "Apple is leading in a smart phone market with iphone sales up by 35%",
        "Tesla booked a minor profit of 1 billion $ in Q2 2022",
        "Apples are high in fiber, Vitamin C and antioxidant",
        "SpaceX got NASA contract worth 10 Billion $"
    ]
)
```

```python
# fetch the documents + embedding

data = collection.get(
    include=["documents", "embeddings", "metadatas"]
)

for i, emb in enumerate(data["embeddings"]):
    print(f"\n Document (i+1)")
    print("Text :", data["documents"][i])
    print("Embedding length:", len(emb))
    print("Embeddings:", emb)
```

```python
collection.query(query_texts="This is related to elon musk", n_results=2)
```
