```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.7-flash")
response = llm.invoke("How many moons are there in jupyter")
print(response)
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.7-flash")
response = llm.invoke([
    ("system", "Give the answer in one line"),
     ("human", "How many moons are in jupyter")
])
print(response.text)
```
